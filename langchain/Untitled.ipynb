```python
# Install the required langchain package
# import subprocess
# import sys

# subprocess.check_call([sys.executable, "-m", "pip", "install", "langchain"])

from langchain_core.prompts import ChatPromptTemplate

prompt = ChatPromptTemplate.from_messages(
    [
        ("system", "You are a helpful assistant that suggests creative restaurant names."),
        ("human", "I want to open a restaurant that serves {cuisine} food. Suggest some fancy name ideas for my restaurant")
    ]
)

prompt.format(cuisine="Italian")

```

```python
from langchain_groq import ChatGroq

llm = ChatGroq(model="llama-3.3-70b-versatile", groq_api_key=groq_api_key)
chain = prompt | llm

response = chain.invoke({"cuisine": "Italian"})
print(response.content)
```
