```python
# Import required libraries
from langchain_core.prompts import ChatPromptTemplate  # For creating prompt templates
from langchain_groq import ChatGroq  # For using Groq's LLM

# Initialize the LLM (Large Language Model)
llm = ChatGroq(
    model="llama-3.3-70b-versatile",  # Model name
    groq_api_key=groq_api_key  # Your API key
)

# STEP 1: Get restaurant name
# Create a prompt template with placeholders using {variable_name}
prompt_name = ChatPromptTemplate.from_messages([
    ("system", "You are a helpful assistant that suggests creative restaurant names."),  # System message sets the AI's role
    ("human", "Suggest one fancy name for a {cuisine} restaurant")  # Human message with {cuisine} placeholder
])

# Create a chain by combining prompt and LLM using pipe operator |
name_chain = prompt_name | llm

# Invoke the chain with actual values for placeholders
name_response = name_chain.invoke({"cuisine": "Indian"})

# Extract the text content from response
restaurant_name = name_response.content
print("Restaurant Name:", restaurant_name)

# STEP 2: Use the restaurant name from Step 1 to get menu items
# Create another prompt template
prompt_items = ChatPromptTemplate.from_messages([
    ("system", "You are a helpful assistant that suggests menu items."),
    ("human", "Suggest menu items for {restaurant_name}. Return as comma separated list")  # Uses {restaurant_name} placeholder
])

# Create another chain
items_chain = prompt_items | llm

# Invoke with the restaurant name we got from Step 1
items_response = items_chain.invoke({"restaurant_name": restaurant_name})

# Print the menu items
print("Menu Items:", items_response.content)
```

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_groq import ChatGroq

# Initialize Groq LLM
llm = ChatGroq(
    model="llama-3.3-70b-versatile",  # Updated model (mixtral-8x7b-32768 is deprecated)
    temperature=0,
    groq_api_key=groq_api_key
)

# Define prompts
prompt_name = ChatPromptTemplate.from_messages([
    ("system", "You are a helpful assistant that suggests creative restaurant names."),
    ("human", "Suggest one fancy name for a {cuisine} restaurant")
])

prompt_items = ChatPromptTemplate.from_messages([
    ("system", "You are a helpful assistant that suggests menu items."),
    ("human", "Suggest menu items for {restaurant_name}. Return as comma separated list")
])

# Create chains
name_chain = prompt_name | llm
food_items_chain = prompt_items | llm

# Execute sequentially (SequentialChain is deprecated, use this approach)
cuisine = "Arabic"

# Step 1: Get restaurant name
name_response = name_chain.invoke({"cuisine": cuisine})
restaurant_name = name_response.content

# Step 2: Get menu items
items_response = food_items_chain.invoke({"restaurant_name": restaurant_name})
menu_items = items_response.content

# Output
result = {
    "cuisine": cuisine,
    "restaurant_name": restaurant_name,
    "menu_items": menu_items
}

print(result)
```
